# un concept tres important dans pytorch: le calcul du gradient

```python
# un neuronne artificielle c'est une approximation d'un neuronne biologique
# pourquoi a t-on besoin d'appliquer une fonction d'activation?
# c'est pour introduire la non linearité
# la majorité des problème sont non linéaires Sigmoïde [0,1], tanh [-1,1], RELU(z)=max(0,z) R+
```

```python
# couche d'entrée, couche cachées, couche de sortie
#Regression (ex: MSE) Classification(ex: entropie croisée)
```

```python
#Données en entrées ->  Passage avant(Forward) <=> Calculs des predictions
# ->Calcul de la perte(LOSS) l(0)
# ->Retropropagation(Backward) <=> Calcul des gradients <=> La derive de la loss par rapport au paramètre W ou au bias
# ->Mise a jour des paramètres 0 <- 0 -nv0L (Algo d'optimisation qui s'appelle la descente du gradient)

#Repeter sur plusieurs iterations(epochs)
```

Un neurone effectue une somme pondérée suivie d'une fonction d'activation <br>
Un réseau de neurones empile plusieurs couches pour apprendre des representation complexes <br>
Les poids et biais sont appris en minimisant une fonction de perte via descente de gradient <br>
Une fois entrainé, le modele peut généraliser à de nouvelle données <br>
La descente de gradient nous permet de trouver les paramètres du modele qui minimise l'erreur <br>
C'est à dire trouver le bon reglage pour rendre l'erreur la plus faible possible

```python
# Comment calculer le gradient avec pytorch
```

```python
#required_grad a chaque fois que je calcule une variable qui depend de moi, 
# calcule automatiquement le gradient
# si je cree une fonction qui depend d'un tenseur avec requires_grad a True le gradient de cette fonction
# sera automatiquement calculer et stockée
# pytorch crée un graph d'operations. et c'est grace à celui ci que pytorch calcule le gradient
# Autrement dit, pytorch suit les operations impliquant ce tenseur.

# quand tu crée les parametre il ne faut pas oublié d'ajouter requires_grad=True, afin de pouvoir 
# calculer les gradients
# on peut acceder à la propriéte grad de chaque paramètre. Qui lui-même est un tenseur.

x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)#pytorch suit les operations impliquant le tenseur w
b = torch.tensor(1.0, requires_grad=True)#pytorch suit les operations impliquant le tenseur b
y =w*x + b
loss =(y-10)**2
loss.backward() # utilise le graphe des opérations pour calculer les gradients
print("y =",y.item)
print("loss =",loss.item())
print("dLoss/dw =",w.grad.item()) # recupère la valeur du gradient dans le tenseurs w
print("dLoss/db =",b.grad.item()) # recupère la valeur du gradient dans le tenseurs b
```

```python
# comment generer avec pytorch le graphe d'operateurs?
# chaque framework a sa fonçon de calculer le gradient
```

// Autre façon de faire


```python
dw,db = torch.autograd.grad(loss,(w,b)) # je veux calculer la deriver du loss par rapport a w et b
print("y =",y.item)
print("loss =",loss.item())
print("dLoss/dw =",dw.item()) 
print("dLoss/db =",db.item()) 
```

```python
x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)#pytorch suit les operations impliquant le tenseur w
b = torch.tensor(1.0, requires_grad=True)#pytorch suit les operations impliquant le tenseur b
y =w*x + b
loss =(y-10)**2

dw,db = torch.autograd.grad(loss,(w,b)) # je veux calculer la deriver du loss par rapport a w et b
print("y =",y.item)
print("loss =",loss.item())
print("dLoss/dw =",dw.item()) 
print("dLoss/db =",db.item()) 
```
